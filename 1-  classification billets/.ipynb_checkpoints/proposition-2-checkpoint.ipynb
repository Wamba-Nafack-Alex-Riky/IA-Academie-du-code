## ANALYSE MULTIVARIÉÉ

```python
#4-faire une reduction a deux composantes principales (PCA) des données
#Préparation du jeu de données pour l'ACP
data_acp=df.drop(columns=['is_genuine'])
#transformation des données en numpy
X=data_acp.values
pca = decomposition.PCA(n_components=2)
#normalisation des données
std_scale = preprocessing.StandardScaler().fit(X)
X_scaled = std_scale.transform(X)
#fit des données numpy
pca.fit(X_scaled)
```

```python
#5-traçage du cercle de correlation 

pcs = pca.components_

fig = plt.subplots(figsize=(10,10))
plt.xlim(-1,1)
plt.ylim(-1,1)

plt.quiver(np.zeros(pcs.shape[1]), np.zeros(pcs.shape[1]),
           pcs[0,:], pcs[1,:], angles='xy', scale_units='xy', scale=1, color='r', width= 0.003)

for i, (x, y) in enumerate(zip(pcs[0, :], pcs[1, :])):
    plt.text(x, y, df.columns[i])
    
circle = plt.Circle((0,0), 1, facecolor='none', edgecolor='b')
plt.gca().add_artist(circle)

# Ajout des axes
plt.plot([-1,1],[0,0],color='silver',linestyle='--',linewidth=1)
plt.plot([0,0],[-1,1],color='silver',linestyle='--',linewidth=1)

plt.title('Cercle des corrélations')
plt.xlabel('F{} ({}%)'.format(1, round(100*pca.explained_variance_ratio_[0],1)))
plt.ylabel('F{} ({}%)'.format(2, round(100*pca.explained_variance_ratio_[1],1)))


plt.show()
```

```python
#6.représentation graphique des donnnées sur les deux composantes 
 
X_projected=pca.transform(X_scaled)
colors = ['red', 'blue']
plt.scatter(X_projected[:, 0], X_projected[:, 1], c=df['is_genuine'].apply(lambda x: colors[x]), 
            )
plt.xlabel('F{} ({}%)'.format(1, round(100*pca.explained_variance_ratio_[0],1)), fontsize=16)
plt.ylabel('F{} ({}%)'.format(2, round(100*pca.explained_variance_ratio_[1],1)), fontsize=16)
plt.title("Projection des {} individus sur les deux composantes".format(X_projected.shape[0]), fontsize=20)
```

#  PARTIE III: MODELISATION ET OPTIMISATION DU MODELE

##  I- MODELISATION

```python
#1-matrices X des données et y targets
X=(df.drop(columns=(['is_genuine']))).values

y=df['is_genuine'].values
```

```python
#2-faire une normalisation de ces données
scaler = preprocessing.StandardScaler()
X = scaler.fit_transform(X)
```

```python
#3-division des données en train et test sets avec 25% des tests
X_train, X_test, y_train, y_test = train_test_split(  X, y, test_size=0.25, random_state=42)
```

```python
#4-entrainnement du modèle
model=LogisticRegression()
model.fit(X_train, y_train)
```

```python
#5-performances du modéle


y_pred=model.predict(X_test)

acc=accuracy_score(y_test, y_pred) 
print(f'notre modèle a une performance de {acc*100}%')
cnf_matrix = confusion_matrix(y_test, y_pred)
print('la matrice de confusion se presente tel que suit:')
cnf_matrix
```

## II-OPTIMISATION

```python
#1-definir un pipeline scikit-learn avec les étapes de normalisation standard et de clasification logistique

model_lo=LogisticRegression()

scaler = preprocessing.StandardScaler()
from sklearn.model_selection import GridSearchCV
model_pipeline = Pipeline([('StandardScaler',scaler), 
                              ('LogisticRegression', model_lo)])
```

```python
#2.créer une grille des hyperparamètres du modèle de classification
hyperparameter= { 
    'C': [1.0, 1.35, 1.75], 
     'solver': ['newtown-cg','lbfgs','liblinear'],
    'penalty': ['l1','l2','elasticnet','none']
         }


gridsearch = GridSearchCV(model_pipeline, 
                          hyperparameter, 
                          scoring='accuracy', 
                          cv=5)
```

```python
gridsearch
```

```python

gridsearch.fit(X_train, y_train)


```

```python
gridsearch.score(X_test, y_test)
```

```python
#4-meilleures valeurs des hyperparamètres choisies par l'optimiseur
print('Tuned hpyerparameters (best parameters):', gridsearch.best_params_)

```

```python
#5-performances du modele
print('Train accuracy:', gridsearch.best_score_)
print('Test accuracy:', gridsearch.best_estimator_.score(X_test, y_test))
```

```python
#8- enregistrement du modèle
#save the trained model
pickle.dump(model, open('regressor.pkl','wb'))
```
